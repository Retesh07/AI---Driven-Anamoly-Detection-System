```python
from google.colab import drive
drive.mount('/content/drive')

import os
import cv2
import numpy as np
import torch
import torch.nn as nn
from tqdm import tqdm
from ultralytics import YOLO
import supervision as sv

device = 0 if torch.cuda.is_available() else "cpu"

print("CUDA available:", torch.cuda.is_available())
print("Using device:", device)
```

```python
pose_model = YOLO("yolov8n-pose.pt")
tracker = sv.ByteTrack()

print("YOLO pose + ByteTrack ready.")
```

```python
def extract_video_features(video_path):

    cap = cv2.VideoCapture(video_path)

    prev_centers = {}
    prev_velocities = {}
    prev_keypoints = {}
    tracker.reset()

    sequence = []
    frame_index = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_index += 1
        if frame_index % 2 != 0:
            continue

        h, w = frame.shape[:2]

        results = pose_model(frame, device=device, conf=0.2, verbose=False)
        result = results[0]

        detections = sv.Detections.from_ultralytics(result)

        if len(detections) == 0:
            sequence.append(np.zeros(87))
            continue

        person_mask = detections.class_id == 0
        detections = detections[person_mask]

        if len(detections) == 0:
            sequence.append(np.zeros(87))
            continue

        detections = tracker.update_with_detections(detections)

        areas = (detections.xyxy[:,2] - detections.xyxy[:,0]) * \
                (detections.xyxy[:,3] - detections.xyxy[:,1])

        order = np.argsort(-areas)
        detections = detections[order]

        person_features = []
        centers = []

        all_velocity_magnitudes = []
        all_keypoints = []

        for i in range(min(len(detections), 2)):

            x1,y1,x2,y2 = detections.xyxy[i]
            track_id = int(detections.tracker_id[i])

            bw = x2-x1
            bh = y2-y1
            if bw <= 0 or bh <= 0:
                continue

            xc = (x1+x2)/2/w
            yc = (y1+y2)/2/h
            centers.append((xc,yc))

            try:
                kps = result.keypoints.xy[i].cpu().numpy()
            except:
                continue

            kp_feats = []
            for (kx,ky) in kps:
                kp_feats.append((kx-x1)/bw)
                kp_feats.append((ky-y1)/bh)

            kp_feats = np.array(kp_feats)
            all_keypoints.append(kp_feats)

            dx = dy = 0
            if track_id in prev_centers:
                px,py = prev_centers[track_id]
                dx = xc - px
                dy = yc - py

            ddx = ddy = 0
            if track_id in prev_velocities:
                pdx,pdy = prev_velocities[track_id]
                ddx = dx - pdx
                ddy = dy - pdy

            velocity_mag = np.sqrt(dx**2 + dy**2)
            all_velocity_magnitudes.append(velocity_mag)

            motion_energy = 0
            if track_id in prev_keypoints:
                motion_energy = np.linalg.norm(
                    kp_feats - prev_keypoints[track_id]
                )

            prev_centers[track_id] = (xc,yc)
            prev_velocities[track_id] = (dx,dy)
            prev_keypoints[track_id] = kp_feats

            bbox_area = bw*bh/(w*h)

            person_feat = np.concatenate([
                kp_feats,
                [dx,dy],
                [ddx,ddy],
                [bbox_area],
                [motion_energy]
            ])

            person_features.append(person_feat)

        while len(person_features) < 2:
            person_features.append(np.zeros(40))

        if len(centers) >= 2:
            (x1,y1),(x2,y2) = centers[:2]
            distance = np.linalg.norm(np.array([x1,y1])-np.array([x2,y2]))
        else:
            distance = 0

        interaction = np.array([distance,0,0])
        scene = np.array([len(detections),0])

        global_motion = np.mean(all_velocity_magnitudes) if len(all_velocity_magnitudes)>0 else 0
        global_kp_variance = np.var(np.vstack(all_keypoints)) if len(all_keypoints)>0 else 0

        frame_feature = np.concatenate([
            person_features[0],
            person_features[1],
            interaction,
            scene,
            [global_motion],
            [global_kp_variance]
        ])

        sequence.append(frame_feature)

    cap.release()

    if len(sequence) >= 60:
        return np.array(sequence[:60])
    else:
        pad = np.zeros((60-len(sequence), 87))
        return np.vstack([sequence, pad])
```

```python
def extract_dataset(input_dir, output_dir):

    for cls in ["Fight", "NonFight"]:
        in_folder = os.path.join(input_dir, cls)
        out_folder = os.path.join(output_dir, cls)
        os.makedirs(out_folder, exist_ok=True)

        files = os.listdir(in_folder)

        for file in tqdm(files):
            video_path = os.path.join(in_folder, file)
            save_path = os.path.join(out_folder, file.replace(".avi", ".npy"))

            features = extract_video_features(video_path)
            np.save(save_path, features)
```

```python
import numpy as np
import os
from tqdm import tqdm

TRAIN_FEATURE_DIR = os.path.join(FEATURE_BASE, "train")

all_features = []

for cls in ["Fight", "NonFight"]:
    folder = os.path.join(TRAIN_FEATURE_DIR, cls)
    files = os.listdir(folder)

    for f in tqdm(files):
        arr = np.load(os.path.join(folder, f))
        all_features.append(arr)

all_features = np.concatenate(all_features, axis=0)

mean = np.mean(all_features, axis=0)
std = np.std(all_features, axis=0)

# 🔥 Stability clamp
std[std < 1e-3] = 1e-3

np.save(os.path.join(FEATURE_BASE, "mean.npy"), mean)
np.save(os.path.join(FEATURE_BASE, "std.npy"), std)

print("Normalization complete.")
print("Feature dimension:", mean.shape)
print("Min std:", np.min(std))
```

```python
import torch
from torch.utils.data import Dataset

class RWFDataset(Dataset):
    def __init__(self, feature_dir, mean, std):
        self.samples = []
        self.mean = mean
        self.std = std

        for label, cls in enumerate(["NonFight", "Fight"]):
            folder = os.path.join(feature_dir, cls)
            for file in os.listdir(folder):
                self.samples.append((os.path.join(folder, file), label))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        features = np.load(path)
        features = (features - self.mean) / self.std
        return torch.tensor(features, dtype=torch.float32), torch.tensor(label, dtype=torch.long)
```

```python
from torch.utils.data import DataLoader

VAL_FEATURE_DIR = os.path.join(FEATURE_BASE, "val")

mean = np.load(os.path.join(FEATURE_BASE, "mean.npy"))
std = np.load(os.path.join(FEATURE_BASE, "std.npy"))

train_dataset = RWFDataset(TRAIN_FEATURE_DIR, mean, std)
val_dataset = RWFDataset(VAL_FEATURE_DIR, mean, std)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

print("Train samples:", len(train_dataset))
print("Val samples:", len(val_dataset))
```

```python
import torch.nn as nn
import torch

device_torch = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class BiLSTMAttention(nn.Module):
    def __init__(self, input_dim=87, hidden_dim=128):
        super().__init__()

        self.lstm = nn.LSTM(
            input_dim,
            hidden_dim,
            batch_first=True,
            bidirectional=True
        )

        self.attn = nn.MultiheadAttention(
            embed_dim=hidden_dim*2,
            num_heads=4,
            batch_first=True
        )

        self.fc = nn.Sequential(
            nn.Linear(hidden_dim*2, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        out, _ = self.lstm(x)
        attn_out, _ = self.attn(out, out, out)
        context = torch.max(attn_out, dim=1)[0]  # MAX pooling
        return self.fc(context)

model = BiLSTMAttention().to(device_torch)

num_fight = len(os.listdir(os.path.join(TRAIN_FEATURE_DIR, "Fight")))
num_nonfight = len(os.listdir(os.path.join(TRAIN_FEATURE_DIR, "NonFight")))

pos_weight = torch.tensor([num_nonfight / num_fight]).to(device_torch)

criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

print("Model initialized.")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

epochs = 25
best_f1 = 0

for epoch in range(epochs):

    model.train()
    train_loss = 0

    for xb, yb in train_loader:
        xb = xb.to(device_torch)
        yb = yb.float().unsqueeze(1).to(device_torch)

        optimizer.zero_grad()
        outputs = model(xb)
        loss = criterion(outputs, yb)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * xb.size(0)

    train_loss /= len(train_loader.dataset)

    model.eval()
    val_loss = 0
    preds_all = []
    targets_all = []

    with torch.no_grad():
        for xb, yb in val_loader:
            xb = xb.to(device_torch)
            yb = yb.float().unsqueeze(1).to(device_torch)

            outputs = model(xb)
            loss = criterion(outputs, yb)

            val_loss += loss.item() * xb.size(0)

            probs = torch.sigmoid(outputs)
            preds = (probs > 0.5).int()

            preds_all.extend(preds.cpu().numpy())
            targets_all.extend(yb.cpu().numpy())

    val_loss /= len(val_loader.dataset)

    precision = precision_score(targets_all, preds_all)
    recall = recall_score(targets_all, preds_all)
    f1 = f1_score(targets_all, preds_all)
    acc = accuracy_score(targets_all, preds_all)

    print(f"\nEpoch {epoch+1}/{epochs}")
    print(f"Train Loss: {train_loss:.4f}")
    print(f"Val Loss:   {val_loss:.4f}")
    print(f"Accuracy:   {acc:.4f}")
    print(f"Precision:  {precision:.4f}")
    print(f"Recall:     {recall:.4f}")
    print(f"F1 Score:   {f1:.4f}")
    print("-"*40)

    if f1 > best_f1:
        best_f1 = f1
        torch.save(model.state_dict(),
                   "/content/drive/MyDrive/RWF2000/best_model_global.pth")
        print("🔥 Best model saved.")
```

```python
from IPython.display import Video, display

def run_inference(video_path, output_video="/content/output_test.mp4"):

    global prev_centers, prev_velocities, prev_keypoints

    prev_centers.clear()
    prev_velocities.clear()
    prev_keypoints.clear()
    tracker.reset()

    temp_dir = "/content/temp_frames"
    os.makedirs(temp_dir, exist_ok=True)

    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps == 0:
        fps = 25

    buffer = deque(maxlen=60)
    prob_history = deque(maxlen=10)

    frame_index = 0
    save_index = 0
    max_prob = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_index += 1
        if frame_index % 2 != 0:
            continue

        frame_feature = extract_frame_feature(frame)
        buffer.append(frame_feature)

        label = "Processing..."
        color = (255,255,0)

        if len(buffer) == 60:

            seq = np.array(buffer)
            seq = (seq - mean) / std

            tensor = torch.tensor(seq, dtype=torch.float32)\
                         .unsqueeze(0)\
                         .to(device_torch)

            with torch.no_grad():
                output = model(tensor)
                prob = torch.sigmoid(output).item()

            prob_history.append(prob)
            smoothed_prob = np.mean(prob_history)

            max_prob = max(max_prob, smoothed_prob)

            if smoothed_prob > 0.5:
                label = f"VIOLENCE {smoothed_prob:.2f}"
                color = (0,0,255)
            else:
                label = f"NORMAL {smoothed_prob:.2f}"
                color = (0,255,0)

        cv2.putText(frame,
                    label,
                    (20,40),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    1,
                    color,
                    2)

        cv2.imwrite(f"{temp_dir}/{save_index:05d}.jpg", frame)
        save_index += 1

    cap.release()

    os.system(
        f"ffmpeg -y -framerate {fps/2} -i {temp_dir}/%05d.jpg "
        f"-c:v libx264 -pix_fmt yuv420p {output_video}"
    )

    print("Max smoothed probability:", round(max_prob,3))
    print("Saved to:", output_video)

    display(Video(output_video, embed=True))
```
